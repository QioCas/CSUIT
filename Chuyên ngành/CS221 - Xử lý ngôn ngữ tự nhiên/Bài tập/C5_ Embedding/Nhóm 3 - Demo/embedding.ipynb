# 1. Load dataset và huấn luyện

```python
from gensim.models import KeyedVectors

import gdown

model = KeyedVectors.load_word2vec_format('GoogleNews-50k.bin', binary=True, limit=500000)
word = input('input your word to show embedding vector: ')
if word in model:
    embedding = model[word]
    print(f"Vector của từ '{word}' (10 chiều đầu):", embedding[:10])
else:
    print("Từ này không có trong từ điển của Google News!")
```

# 2. Thực hiện ví dụ 'king-male+female'

```python
result=model.most_similar(positive=['Messi', 'Female'], negative=['Male'], topn=3)
for word, similarity in result:
    print(f'{word}: {similarity: .4f}')
```

### Trường hợp tổng quát của phần 2

```python
pos=list(input().split())
neg=list(input().split())
result=model.most_similar(positive=pos, negative=neg, topn=3)
for word, score in result:
    print(f"{word}: {score: .5f}")
```

# 3. Tìm từ đồng nghĩa

```python
w1=input("input your word: ")
result=model.most_similar(w1, topn=3)
for word, score in result:
    print(f'{word}: {score: .4f}')
```

# 4. Kiểm tra sự tương quan giữa 2 từ

```python
w1, w2 =input().split()
print(model.similarity(w1,w2))
```

# 5. Tìm từ khác loại

```python
w=['apple','orange','lemon','sport','watermelon']
uni_word=model.doesnt_match(w)
print(uni_word)
```

# 6. Minigame: đoán tên thủ đô

```python
def predict_capital(country):
    result=model.most_similar(positive=['Tehran', country], negative=['Iran'], topn=1)
    return result[0][0], result[0][1]
country=input('Nhap ten quoc gia: ').capitalize()
word, score=predict_capital(country)
print(f'{word}: {score: .5f}')
```
