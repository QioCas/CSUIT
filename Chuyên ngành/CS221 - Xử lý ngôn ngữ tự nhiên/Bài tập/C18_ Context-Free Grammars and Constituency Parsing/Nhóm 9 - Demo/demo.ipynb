```python
def cky_parse(words, grammar):
    n = len(words)
    # Khởi tạo bảng n x n chứa các tập hợp (set) rỗng
    # table[i][j] sẽ lưu các biến (Non-terminals) có thể tạo ra chuỗi từ words[i] đến words[j]
    table = [[set() for _ in range(n)] for _ in range(n)]

    # Bước 1: Xử lý các từ đơn lẻ (Điền đường chéo chính của bảng)
    # Áp dụng các luật dạng: A -> a (Ví dụ: NP -> "mèo")
    for i, word in enumerate(words):
        for lhs, rhs_list in grammar.items():
            for rhs in rhs_list:
                if len(rhs) == 1 and rhs[0] == word:
                    table[i][i].add(lhs)

    # Bước 2: Xử lý các chuỗi con dài hơn theo độ dài tăng dần
    # Áp dụng các luật dạng: A -> B C
    for length in range(2, n + 1):        # length: độ dài của chuỗi con đang xét
        for i in range(n - length + 1):   # i: vị trí bắt đầu
            j = i + length - 1            # j: vị trí kết thúc
            
            # Thử chia chuỗi con tại mọi vị trí k (từ i đến j-1)
            for k in range(i, j):
                # Nếu B có thể tạo ra phần nửa đầu (i đến k) 
                # và C có thể tạo ra phần nửa sau (k+1 đến j)
                for lhs, rhs_list in grammar.items():
                    for rhs in rhs_list:
                        if len(rhs) == 2:
                            B, C = rhs[0], rhs[1]
                            if B in table[i][k] and C in table[k + 1][j]:
                                table[i][j].add(lhs)

    return table
```

```python
# 1. Định nghĩa ngữ pháp (Dạng chuẩn Chomsky - CNF)
# S: Câu, NP: Cụm danh từ, VP: Cụm động từ, V: Động từ
grammar = {
    'S': [['NP', 'VP']],
    'VP': [['V', 'NP']],
    'NP': [['mèo'], ['cá'], ['chó']],
    'V': [['ăn'], ['cắn']]
}

# 2. Câu cần phân tích
sentence = "mèo ăn cá"
words = sentence.split()

# 3. Chạy thuật toán CKY
print(f"Phân tích câu: '{sentence}'\n" + "-"*30)
table = cky_parse(words, grammar)

# 4. In bảng kết quả
for length in range(1, len(words) + 1):
    for i in range(len(words) - length + 1):
        j = i + length - 1
        phrase = " ".join(words[i:j+1])
        tags = list(table[i][j]) if table[i][j] else "Không có"
        print(f"[{i},{j}] '{phrase}': {tags}")

# 5. Kết luận
print("-" * 30)
if 'S' in table[0][len(words)-1]:
    print("=> Kết luận: CÂU HỢP LỆ! (Có thể được dẫn xuất từ ký hiệu bắt đầu S).")
else:
    print("=> Kết luận: CÂU KHÔNG HỢP LỆ!")
```
